```python
from sklearn.model_selection import KFold
import numpy as np
```

```python
n_samples = 20
dim = 10

def get_num_observations_in_train_split(kf: KFold, split_idx: int, n_observations: int):
    ratio = float(n_observations) / float(kf.n_splits)
    num_splits_with_less_than_max_test_size = kf.n_splits * np.ceil(ratio) - n_observations
    if split_idx < kf.n_splits - num_splits_with_less_than_max_test_size:
        return n_observations - np.ceil(ratio)
    else:
        return n_observations - np.floor(ratio)

for j in range(100):
    generator = np.random.RandomState(j)
    data = np.random.rand(n_samples, dim)
    kf = KFold(n_splits=5, shuffle=True, random_state=generator)
    ratio = float(n_samples) / float(kf.n_splits)
    first_split_test_size = np.ceil(ratio)
    for i, (train_index, test_index) in enumerate(kf.split(data)):
        num_splits_with_less_than_max_test_size = kf.n_splits * first_split_test_size - n_samples
        training_set_size = get_num_observations_in_train_split(kf, i, n_samples)
        assert training_set_size == len(train_index)
        assert len(test_index) == n_samples - training_set_size
```
